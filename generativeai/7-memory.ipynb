### Agent with memory

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite"
)

agent = create_agent(
    model=model,
    checkpointer=InMemorySaver()
    
)
```

```python
### Run with thread id
config={"configurable":{"thread_id":"test1"}}
```

```python
from langchain.messages import HumanMessage,SystemMessage

response=agent.invoke({"messages":[HumanMessage(content="hi i am aryan")]},config=config)
response
```

```python
agent.invoke({"messages":[HumanMessage(content="Hi what is my name")]},config=config)
```

```python
### Run with thread id for another user
config={"configurable":{"thread_id":"test2"}}

from langchain.messages import HumanMessage,SystemMessage

response=agent.invoke({"messages":[HumanMessage(content="hi i am luffy")]},config=config)
response
```
