### Generative AI Application with LLM models

```python
from langchain.chat_models import init_chat_model
from langchain_google_genai import ChatGoogleGenerativeAI
#init_chat_model can be used to initialize any chat model while ChatGoogleGenerativeAI is for gemini Models same can be used for ChatOpenAI instead of init_chat_model

# gemini = ChatGoogleGenerativeAI(
#     model="gemini-3.6-flash",
#     temperature=0
# )
gemini = init_chat_model("google_genai:gemini-3.6-flash",temperature=0)
```

```python
response = gemini.invoke("What is ai agent")
response
```

### Streaming: Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses.

```python
for chunk in gemini.stream("What is Existensialism and existensial crisis"):
    print(chunk.text,end = " ",flush=True )
```

### Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:

```python
responses = gemini.batch([
    "What is Abusrdism?",
    "What is password of elon musk private holdings?",
    "What is existentialism?"
])
for response in responses:
    print(response)
```
