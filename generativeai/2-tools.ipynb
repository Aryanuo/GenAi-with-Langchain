## Tools: 
Models can request to call tools that perform tasks such as fetching data from a dataase, searching the web or running code. Tools are pairings of:

1. A schema, including the name of the tool a description, and/or argument definitions (often a JSON schema)
2. A function of coroutine to execute

*Tools extend what agents can do—letting them fetch real-time data, execute code, query external databases, and take actions in the world.* 

```python
import langchain
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")

model = init_chat_model("google_genai:gemini-3.5-flash-lite",temperature=0)
```

```python
response = model.invoke("What is ai agent")
response
```

```python
# TOOLS

from langchain.tools import tool
@tool
def get_weather(location:str)->str:
    """Get the weather at location"""
    return f"It's sunny in {location }"

model_with_tools=model.bind_tools([get_weather])

```

```python
response=model_with_tools.invoke("What is the weather in India")
print(response)
```

```python
response.tool_calls
```

### Tool Execution Loop

```python
# Step 1: Model generates tool calls
messages = [{"role": "user", "content": "what's the weather in India"}]
ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)
print(ai_msg)
print(" ------ ")
print(messages)
print(" ---- ")

# Step 2: Execute Tools and collect results
for tool_call in ai_msg.tool_calls:
    print(tool_call)
    print(" ---- ")
    # execute the tool with the generated arguments
    tool_result = get_weather.invoke(tool_call)
    print(tool_result)
    print(" ---- ")
    messages.append(tool_result)
    print(messages)
    print(" ---- ")


# Step 3: Pass results back to model for final response
final_response = model_with_tools.invoke(messages)
print(final_response)


```
